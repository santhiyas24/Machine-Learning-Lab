```python
import numpy as np
from scipy import linalg
A = np.array([[1, 2],
              [3, 4]])
print("Matrix A:")
print(A)
print("\nDeterminant:")
print(linalg.det(A))
print("\nInverse:")
print(linalg.inv(A))
print("\nTranspose:")
print(A.T)
```

```python
import numpy as np
from scipy import linalg
A = np.array([[1,2],
              [3,4]])
eigen_values, eigen_vectors = linalg.eig(A)
print("Eigen Values:")
print(eigen_values)
print("\nEigen Vectors:")
print(eigen_vectors)
```

```python
from scipy.stats import chisquare
observed = [40,10,20,30]
expected = [30,20,30,20]
chi, p = chisquare(f_obs=observed,
                   f_exp=expected)
print("Chi-Square Statistic =", chi)
print("P-value =", p)
```
